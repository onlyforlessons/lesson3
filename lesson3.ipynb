# TASK 1    Grade system

```python
student_score = (float(input("Enter your score: ")))
if student_score >= 91:
    print(f"{student_score} --> Your grade is A")
elif student_score >= 81:  
    print(f"{student_score} --> Your grade is B")
elif student_score >= 71:
    print(f"{student_score} --> Your grade is C")
elif student_score >= 61:
    print(f"{student_score} --> Your grade is D")    
elif student_score >= 51:
    print(f"{student_score} --> Your grade is E")    
else:
    print(f"{student_score} --> Your grade is F")
```

#   Task 2   Multipication Table

```python
number = int(input("Enter a number: "))

for i in range(1, 11):
    result = number * i
    print(f"{number} x {i} = {result}") 

```

#   Task 3 Password retry system 

```python
correct_password = "python123"
user_password = input("Enter your password: ")
attempts = 0   # attempts limit is set to 3 
while user_password != correct_password : 
    attempts += 1
    if attempts >= 3:
        print("Too many incorrect attempts. Access denied.")
        break
    user_password = input("Incorrect password. Please try again: ")
```
